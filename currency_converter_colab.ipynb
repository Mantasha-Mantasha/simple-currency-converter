# Simple Currency Converter
A beginner-friendly Python project. It fetches the latest available rate from a public endpoint without requiring an API key. If the live service is unavailable, you can enter a rate manually.

**Example:** convert 10 USD to PKR. Rates may differ from banks and money exchangers.

```python
import json
import urllib.request
import urllib.error

def get_rates(base_currency):
    url = f"https://open.er-api.com/v6/latest/{base_currency.upper()}"
    request = urllib.request.Request(url, headers={"User-Agent": "SimpleCurrencyConverter/1.0"})
    try:
        with urllib.request.urlopen(request, timeout=10) as response:
            data = json.loads(response.read().decode("utf-8"))
        if data.get("result") == "success":
            return data.get("rates", {})
    except (urllib.error.URLError, TimeoutError, json.JSONDecodeError) as exc:
        print("Could not retrieve live rates:", exc)
    return None

def convert(amount, rate):
    return amount * rate
```

```python
print("SIMPLE CURRENCY CONVERTER")
print("Example codes: PKR, USD, EUR, GBP, SAR, AED")
source = input("Convert FROM (e.g., USD): ").strip().upper()
target = input("Convert TO (e.g., PKR): ").strip().upper()

if not source or not target or len(source) != 3 or len(target) != 3:
    print("Please enter valid 3-letter currency codes.")
else:
    try:
        amount = float(input(f"Amount in {source}: "))
        if amount < 0:
            raise ValueError("Amount cannot be negative")
        rates = get_rates(source)
        if rates and target in rates:
            rate = rates[target]
            print("Using latest available public exchange rate.")
        else:
            print("Live rate unavailable. Enter the target units for 1 source unit.")
            rate = float(input(f"1 {source} = how many {target}? "))
            if rate < 0:
                raise ValueError("Rate cannot be negative")
            print("Using your manually entered rate.")
        result = convert(amount, rate)
        print(f"\\n{amount:,.2f} {source} = {result:,.2f} {target}")
        print(f"Rate used: 1 {source} = {rate:,.6f} {target}")
        print("Note: bank rates and fees may differ.")
    except ValueError as exc:
        print("Input error:", exc)
```

## How it works
1. The user selects a source and target currency.
2. The program requests rates for the source currency.
3. It selects the target currency's rate.
4. It calculates: **Converted amount = Amount × Exchange rate**.
5. If live rates cannot be retrieved, the user can enter a rate manually.
